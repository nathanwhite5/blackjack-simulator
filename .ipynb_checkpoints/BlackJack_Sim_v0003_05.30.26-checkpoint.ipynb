```python
import pandas as pd
import random as rd
import numpy as np

# Deck Builder
suits_list = [
    'Spades'
    ,'Hearts'
    ,'Clubs'
    ,'Diamonds'
]

rank_list = [
    'Ace'
    ,'King'
    ,'Queen'
    ,'Jack'
    ,'Ten'
    ,'Nine'
    ,'Eight'
    ,'Seven'
    ,'Six'
    ,'Five'
    ,'Four'
    ,'Three'
    ,'Two'
]

single_deck = []

for suit in suits_list:
    for rank in rank_list:
        single_deck.append(f'{rank} {suit}')

# not necessary for final run, delete
#len(single_deck)
#type(single_deck)
#print(single_deck)

# Value Assigner (for later)
##note, not sure how to handle ace 11/1 switcher
## choosing Ace = 11 for now
rank_value_dict = {'Ace': 11
                   ,'King':10
                   ,'Queen':10
                   ,'Jack':10
                   ,'Ten':10
                   ,'Nine':9
                   ,'Eight':8
                   ,'Seven':7
                   ,'Six':6
                   ,'Five':5
                   ,'Four':4
                   ,'Three':3
                   ,'Two':2
}

#note, live_deck doesnt work within the function
live_deck = rd.sample(single_deck, len(single_deck))

coin_purse = 50

## playing single hand
def play_hand():

    print(f'\n')
    print(f'There are currently {len(live_deck)} cards left in the deck')

    max_total = 21
    
    ##player hand
    player_hand = []
    player_hand_total = 0
    player_hand_size = 0

    ##dealer hand
    dealer_hand = []
    dealer_hand_total = 0
    dealer_hand_size = 0

    ## INITIAL DEAL
    #first player card draw
    player_hand.append(live_deck.pop())
    #player_hand_size = player_hand_size + 1

    #first dealer card draw
    dealer_hand.append(live_deck.pop())
    #dealer_hand_size = dealer_hand_size + 1

    #second player card draw
    player_hand.append(live_deck.pop())
    #player_hand_size = player_hand_size + 1

    #second dealer card draw
    dealer_hand.append(live_deck.pop())
    #dealer_hand_size = dealer_hand_size + 1

    ## list comprehend that shit yo!
    dealer_face = []
    dealer_face.append(dealer_hand[0])
    
    #needs to call Dealer's Blackjack Immediately
    if dealer_hand_total == 21:
        print(f'Blackjack!')
        return
        
    AcePlay_Player = False
    drop_10_toggle_player = False
    #player card value assign
    for card in player_hand:
        rank,suit = card.split()
        player_hand_total = player_hand_total + rank_value_dict[rank]
        if rank == 'Ace':
            AcePlay_Player = True


    AcePlay_Dealer = False
    drop10_toggle_dealer = False
    #dealer card value assign
    for card in dealer_hand:
        rank,suit = card.split()
        dealer_hand_total = dealer_hand_total + rank_value_dict[rank]
        if rank == 'Ace':
            AcePlay_Dealer = True

    #print(f"Player's Current Hand: {player_hand}. Hand Total: {player_hand_total}")
    #print(f"Dealer Showing: {dealer_hand[0]}. Dealer Showing Total: {rank_value_dict[dealer_hand[0].split()[0]]}")
    #print(f"Number of Cards Remaining in Deck: {len(live_deck)}")

    if player_hand_total == 21:
        print(f'\n')
        print(f'Player has {player_hand} for a total of {player_hand_total}. Dealer has {dealer_hand_total} for a total of {dealer_hand_total}.')
        if dealer_hand_total == 21:
            print(f'\n')
            print(f'Draw! Both Player and Dealer have Blackjack!')
            return f'draw'
        print(f'\n')       
        print(f'Player Wins! Player has Blackjack!')
        return f'win'
    elif dealer_hand_total == 21:
        print(f'\n')
        print(f'Player has {player_hand} for a total of {player_hand_total}. Dealer has {dealer_hand_total} for a total of {dealer_hand_total}.')
        print(f'\n')
        print(f'Dealer wins! Dealer has Blackjack')
        return f'lose'

            
    ##user input
    # '1' and '0' are strings
    user_data = '1'
    drop_10_toggle_player = False
    
    while user_data != '0':

        #note, dont know how to code in space in print, temporary solution in another print
        print(f'\n')
        print(f"You have {player_hand} for a total of {player_hand_total}.")  
        print(f"\n")
        print(f"Dealer is showing {dealer_face} for a total of {rank_value_dict[dealer_hand[0].split()[0]]}.")
        print(f"\n")
        print(f"Press '1' to HIT, Press '0' to STAND")
        user_data = input()
        
        if user_data == '1':

            print(f'\n')
            print(f'Player HITS')
            
            #draw card function, embed in deal hand function inside 'hit'
            #ensure player_hand_total also updates
            #ensure total cards left updates
            new_card = live_deck.pop()
            
            player_hand.append(new_card)

            rank,suit = new_card.split()

            print(f'\n')
            print(f'You have drawn the {rank} of {suit}')
            
            player_hand_total = player_hand_total + rank_value_dict[rank]
            
            if rank == 'Ace':
                AcePlay_Player = True
        
            if player_hand_total > 21:
                #print(f'player hand > 21')
                if AcePlay_Player == True:
                    if drop_10_toggle_player == False: #include single-use drop10 trigger
                        player_hand_total = player_hand_total - 10
                        drop_10_toggle_player = True
                    
                    else:
                        print(f'\n')
                        print(f'Player Bust! Player hand is {player_hand} for a total of {player_hand_total}')
                        return f'lose'
                else:
                    print(f'\n')
                    print(f'Player goes Boom! Player hand is {player_hand} for a total of {player_hand_total}')
                    return f'lose'
            ###
        elif user_data == '0':
            print(f'\n')
            print(f'Player STANDS')
            ###
        else:
            print(f'\n')
            print(f"INCORRECT ENTRY")
            print(f'\n')


    ## dEALER TINGS
    drop_10_toggle_dealer = False

    # redundant?
    #if dealer_hand_total == 21:
    #    print(f'Blackjack!')
        
    #greater 17
    if dealer_hand_total > 17:
        print(f'\n')
        print(f'Dealer Stands')
        if dealer_hand_total > player_hand_total:
            print(f'\n')
            print(f'Dealer shows {dealer_hand} for a total of {dealer_hand_total}, which beats your {player_hand_total}')
            return f'lose'
        elif dealer_hand_total < player_hand_total:
            print(f'\n')
            print(f'Dealer shows {dealer_hand} for a total of {dealer_hand_total}, but your {player_hand_total} wins!')
            return f'win'
        else:
            print(f'\n')
            print(f'Draw! Dealer shows {dealer_hand} for a total of {dealer_hand_total}, which is equal to your {player_hand_total}!')
            return f'draw'
            

    #equal 17
    if dealer_hand_total == 17: #note, hit on soft 17'
        print(f'\n')
        print(f'equaller')
        
        if AcePlay_Dealer == False:
            if dealer_hand_total > player_hand_total:
                print(f'\n')
                print(f'Dealer Stands')
                print(f'\n')
                print(f'Dealer shows {dealer_hand} for a total of {dealer_hand_total}, which beats your {player_hand_total}')
                return f'lose'
            elif dealer_hand_total < player_hand_total:
                print(f'\n')
                print(f'Dealer Stands')
                print(f'\n')
                print(f'Dealer shows {dealer_hand} for a total of {dealer_hand_total}, but your {player_hand_total} wins!')
                return f'win'
            else:
                print(f'\n')
                print(f'Dealer Stands')
                print(f'\n')
                print(f'Draw! Dealer shows {dealer_hand} for a total of {dealer_hand_total}, which is equal to your {player_hand_total}!')
                return f'draw'
        else:
            
            player_hand_total = player_hand_total - 10
            
            drop_10_player = True
            
            new_card = dealer_hand.append(live_deck.pop())
        
            dealer_hand.append(new_card)
    
            rank,suit = new_card.split()
        
            dealer_hand_total = dealer_hand_total + rank_value_dict[rank]

            print(f'The Dealers hand is now: {dealer_hand}')
    
    #less 17
    while dealer_hand_total < 17:

        new_card = live_deck.pop()

        print(f'\n')
        print(f'Dealer draws {new_card}')
        print(f'\n')
        
        dealer_hand.append(new_card)
    
        rank,suit = new_card.split()     
        
        dealer_hand_total = dealer_hand_total + rank_value_dict[rank]
        
        if rank == 'Ace':
            AcePlay_Dealer = True

        if dealer_hand_total > 21:
            print(f'Dealer Busts! You Win!')
            return f'win'
        print(f'The Dealers hand is now {dealer_hand} for a total of {dealer_hand_total}')

    if dealer_hand_total > player_hand_total:
        print(f'\n')
        print(f'Dealer shows {dealer_hand} for a total of {dealer_hand_total}, which beats your {player_hand_total}')
        return f'lose'
    elif dealer_hand_total < player_hand_total:
        print(f'\n')
        print(f'Dealer shows {dealer_hand} for a total of {dealer_hand_total}, but your {player_hand_total} wins!')
        return f'win'
    else:
        print(f'\n')
        print(f'Draw! Dealer shows {dealer_hand} for a total of {dealer_hand_total}, which is equal to your {player_hand_total}!')
        return f'draw'




        
# **********************************
# **********************************
# *********Play Game Here********
# **********************************
# **********************************

def play_game():

    #live_deck_len = len(live_deck)
    #while len(live_deck) >= live_deck_len/2:
    # previously '1' = hit, '0' = stand
    # note, need for both upper case and lower case

    # give the while-loop something to chew on (establish variable for while-loop)
    player_choice = 'y'

    while player_choice != 'n':
        play_hand() #function above
        print(f'\n')
        print(f'Player select Y to continue playing, N to stop')
        player_choice = input() 
        
#    if user_data == 'n':
#        break
#    else:
#        play_hand()

    
```
